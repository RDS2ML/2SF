## **library BeautifulSoup**

```python
from bs4 import BeautifulSoup
import requests
from pprint import pprint
```

#### **GET запрос**

```python
url = 'https://nuancesprog.ru/p/11460/'
response = requests.get(url)
# создаем объект BS 
page = BeautifulSoup(response.text, 'html.parser')
page.text
```

#### **Извлекаем заголовок статьи**

```python
# Применяем метод find() к объекту и выводим результат на экран
print(page.find('h1').text)
```

#### **Неуникальные теги**

```python
pprint(page.find('p').text) # первый встретившийся параграф 
[p for p in page.find_all('p')] # все праграфы на странице 
```
