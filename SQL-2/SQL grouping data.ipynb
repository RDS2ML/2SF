## **SQL grouping data\SQL группировка данных**

#### **GROUP BY**

* SELECT
    * type1 AS pokemon_type,
    * COUNT(*) AS pokemon_count
* FROM sql.pokemon
* GROUP BY type1
* ORDER BY type1
